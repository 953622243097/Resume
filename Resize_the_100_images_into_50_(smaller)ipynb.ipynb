```python
import cv2
import os
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
output_dir = '/content/output_images'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

```

```python
image_files = list(uploaded.keys())

for image_file in image_files:
    # Read the image
    image = cv2.imread(image_file)

    # Check if the image was read successfully
    if image is None:
        print(f"Error reading {image_file}. Skipping.")
        continue

    # Get the dimensions of the image
    height, width = image.shape[:2]

    # Calculate the new dimensions
    new_dimensions = (width // 2, height // 2)

    # Resize the image
    resized_image = cv2.resize(image, new_dimensions, interpolation=cv2.INTER_AREA)

    # Construct the full output path
    output_path = os.path.join(output_dir, image_file)

    # Save the resized image
    cv2.imwrite(output_path, resized_image)

    print(f"Resized {image_file} and saved to {output_path}")
```

```python
import shutil

shutil.make_archive('resized_images', 'zip', output_dir)
files.download('resized_images.zip')
```
