```python
class Student:
    def __init__(self, name):
        self.name = name
        self.marks = []

    def add_mark(self, mark):
        self.marks.append(mark)

    def get_average(self):
        if not self.marks:
            return 0
        return sum(self.marks) / len(self.marks)


class MarksManager:
    def __init__(self):
        self.students = {}

    def add_student(self, name):
        if name not in self.students:
            self.students[name] = Student(name)
            print(f"Student '{name}' added.")
        else:
            print(f"Student '{name}' already exists.")

    def add_mark(self, name, mark):
        if name in self.students:
            self.students[name].add_mark(mark)
            print(f"Mark {mark} added to student '{name}'.")
        else:
            print(f"Student '{name}' does not exist. Please add the student first.")

    def display_students(self):
        for name, student in self.students.items():
            print(f"Student: {name}, Marks: {student.marks}")

    def get_student_average(self, name):
        if name in self.students:
            avg = self.students[name].get_average()
            print(f"Average marks for student '{name}': {avg:.2f}")
        else:
            print(f"Student '{name}' does not exist.")

    def get_highest_and_lowest(self):
        if not self.students:
            print("No students in the system.")
            return

        all_marks = [mark for student in self.students.values() for mark in student.marks]

        if not all_marks:
            print("No marks available to find highest and lowest.")
            return

        highest = max(all_marks)
        lowest = min(all_marks)
        print(f"Highest mark: {highest}")
        print(f"Lowest mark: {lowest}")


def main():
    manager = MarksManager()

    while True:
        print("\nMenu:")
        print("1. Add Student")
        print("2. Add Mark")
        print("3. Display All Students")
        print("4. Get Student Average")
        print("5. Get Highest and Lowest Marks")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == '1':
            name = input("Enter student name: ")
            manager.add_student(name)
        elif choice == '2':
            name = input("Enter student name: ")
            try:
                mark = float(input("Enter mark: "))
                manager.add_mark(name, mark)
            except ValueError:
                print("Please enter a valid number for the mark.")
        elif choice == '3':
            manager.display_students()
        elif choice == '4':
            name = input("Enter student name: ")
            manager.get_student_average(name)
        elif choice == '5':
            manager.get_highest_and_lowest()
        elif choice == '6':
            break
        else:
            print("Invalid choice. Please try again.")


if __name__ == "__main__":
    main()
```
